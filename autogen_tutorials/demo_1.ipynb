```python
import os
import autogen
from dotenv import load_dotenv

customer_agent = autogen.UserProxyAgent(
    name="Customer",
    human_input_mode="ALWAYS",
    code_execution_config={"use_docker": False},
    max_consecutive_auto_reply=2,
)
```

```python
support_agent = autogen.AssistantAgent(
    name="SupportAgent",
    llm_config={
        "config_list": [
            {
                "model": "gpt-4o-mini",
                "api_key": os.getenv("OPENAI_API_KEY")
            }
        ],
        "temperature":0.7
    },
    system_message= "You are a helpful AI support agent. Answer customer queries clearly and concisely. If you don't know the answer, say you don't know.",
    code_execution_config={"use_docker": False},
    max_consecutive_auto_reply=2,
)
```

```python
# Step-3: Running a simulated Customer Intraction

customer_agent.initiate_chat(
    support_agent,
    message="I'm experiencing issue with my account. can you help me to resolve it?"
)
```
